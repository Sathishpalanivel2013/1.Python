```python
#Random Forest regressor
# 1. Import necessary library 
import pandas as pd
```

```python
# 2. Load Dataset
Dataset = pd.read_csv("insurance_pre.csv")
```

```python
Dataset
```

```python
# 3. Input Output split
Independent = Dataset.drop(columns=["charges"])
dependent = Dataset["charges"]
```

```python
Independent
```

```python
dependent
```

```python
#4 convert catagorical data
Independent = pd.get_dummies(Independent,drop_first=True)
```

```python
Independent
```

```python
# 5 Train test split
from sklearn.model_selection import train_test_split 
X_train,X_test,y_train,y_test= train_test_split( Independent,dependent, test_size=0.03, random_state=0)
```

```python
# 6 Scaling the feature
from sklearn.preprocessing import StandardScaler
sc = StandardScaler( )
X_train_scaled = sc.fit_transform(X_train)
X_test_scaled = sc.transform(X_test)
```

```python
# 5 Model_selection & cross_validation
from sklearn.ensemble import RandomForestRegressor
RF = RandomForestRegressor( criterion = 'squared_error' ,n_estimators = 200 )
RF.fit(X_train_scaled,y_train)
```

```python
#6 Viewing weight and bias value
#weight= DR.coef_
#print(f"Weight of the model { weight }")
#Bias = DR.intercept_
#print(f"Bias of the model { Bias }")
```

```python
# 7 Model_Evalution & Digonestices
Y_predict = RF.predict(X_test_scaled)
```

```python
Y_predict
```

```python
# 8 Check r_score value performance of the Model
from sklearn.metrics import r2_score
r_score = r2_score(y_test,Y_predict)
r_score
```

```python
#9 Deployment model to file
import pickle
pickle.dump(RF,open("RF_regress_insur.svc","wb"))
model = pickle.load(open("RF_regress_insur.svc","rb"))
```
